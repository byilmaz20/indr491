```python
import pandas as pd

# Excel dosyasını oku
df = pd.read_excel("../preprocessedBelgeler/NewforecastData_altLimit100_son3_byGrup.xlsx")

# Kolon isimlerini düzenle
df.columns = df.columns.str.strip().str.replace(" ", "_")
df.rename(columns={"Sipariş_Mik._(TON)": "Sipariş_Mik", "Month": "Ay"}, inplace=True)

# Ay kolonunu datetime'a çevir
df['Ay'] = pd.to_datetime(df['Ay'])

# SpecGroupId’ye göre gruplama ve analiz
spec_summary = (
    df.groupby('SpecGroupId')
      .agg(
          Total_Siparis_Miktari=('Sipariş_Mik', 'sum'),
          Farkli_Ay_Sayisi=('Ay', lambda x: x.dt.to_period('M').nunique())
      )
      .sort_values(by='Total_Siparis_Miktari', ascending=False)
      .head(10)
)

print(spec_summary)
```
