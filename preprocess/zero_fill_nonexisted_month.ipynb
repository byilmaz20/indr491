```python
import pandas as pd
import random

# Dosyayı oku
file_path = "/Users/cankocak/Desktop/Spring25/INDR491/indr491/preprocessedBelgeler/time_formatted_forecastData_altLimit100_son3_byGrup.xlsx"  # Gerçek dosya yolunu yaz
df = pd.read_excel(file_path)


# Tarih formatını düzenle
df["Month"] = pd.to_datetime(df["Month"], format="%Y%m%d")

# Tüm unique_id'ler için eksik ayları tamamla
all_months = pd.date_range(start=df["Month"].min(), end=df["Month"].max(), freq='MS')

# Yeni satırları saklayacak liste
new_rows = []

# Her unique_id için eksik ayları kontrol et ve ekle
for uid in df["SpecGroupId"].unique():
    existing_months = df[df["SpecGroupId"] == uid]["Month"].unique()
    missing_months = [m for m in all_months if m not in existing_months]
    grade = df.loc[df["SpecGroupId"] == uid, "Grade"].unique()[0]
    for month in missing_months:
        new_rows.append({"SpecGroupId": uid, "Month": month,"Grade": grade ,"Sipariş Mik. (TON)": 0})  # "x" için varsayılan değer

# Yeni satırları ekle
df = pd.concat([df, pd.DataFrame(new_rows)], ignore_index=True)


# CSV veya Excel'e kaydetmek için (isteğe bağlı)
df.to_excel("/Users/cankocak/Desktop/Spring25/INDR491/indr491/preprocessedBelgeler/zero_filled_forecastData_altLimit100_son3_byGrup.xlsx.xlsx", index=False)
```
